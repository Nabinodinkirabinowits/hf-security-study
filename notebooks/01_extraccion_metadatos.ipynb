# 01 - Extracción y Análisis Preliminar de Datasets de Seguridad (Hugging Face)

Este notebook consulta la API de Hugging Face a través de `HfApi` para recolectar metadatos de los datasets etiquetados con tópicos de seguridad y ciberseguridad, analizando patrones de uso, descargas y fechas.

```python
import sys
from pathlib import Path

# Agregar raíz al path
ROOT_DIR = Path.cwd().parent
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.extractor import HFSecurityExtractor

sns.set_theme(style="whitegrid")
```

## 1. Extracción de Metadatos
Descargamos metadatos de datasets con tags `security`, `cybersecurity`, `vulnerability`.

```python
extractor = HFSecurityExtractor()
df = extractor.fetch_multi_tag(tags=["security", "cybersecurity", "vulnerability"], limit_per_tag=300)
saved_paths = extractor.save_metadata(df)
print(f"Total datasets únicos obtenidos: {len(df)}")
df.head()
```

## 2. Top 15 Datasets por Descargas Recientes (Últimos 30 días)

```python
top_downloads = df.sort_values(by="downloads_30d", ascending=False).head(15)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_downloads, x="downloads_30d", y="id", palette="viridis")
plt.title("Top 15 Datasets de Seguridad por Descargas (30 días)")
plt.xlabel("Descargas (Últimos 30 días)")
plt.ylabel("Dataset ID")
plt.tight_layout()
plt.show()
```

## 3. Correlación: Descargas (30d) vs Likes
Comparación para observar si la popularidad estática (likes) refleja la adopción real activa (downloads).

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="likes", y="downloads_30d", alpha=0.7)
plt.xscale("log")
plt.yscale("log")
plt.title("Likes vs Descargas Recientes (Escala Logarítmica)")
plt.xlabel("Likes (log)")
plt.ylabel("Downloads 30d (log)")
plt.tight_layout()
plt.show()
```
